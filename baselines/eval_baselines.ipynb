```python
import pandas as pd
from sklearn.metrics import classification_report, accuracy_score

for lang in ["english", "hindi", "chinese", "all"]:    
    df = pd.read_csv(f"{lang}_test_baseline.csv")
    true_df = pd.read_excel(f"../data/{lang}/Test_labels.xlsx")
    y_true = true_df["label"].tolist()
    y_pred = df["label"].tolist()
    print(f"Results for {lang}:")
    print(classification_report(y_true, y_pred))
```
